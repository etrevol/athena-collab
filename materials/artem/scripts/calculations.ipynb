```python
# NGC 1068

import numpy as np

# === ФІЗИЧНІ КОНСТАНТИ ===
G     = 6.67430e-8       # гравітаційна стала [см^3 г^-1 с^-2]
c     = 2.99792458e10    # швидкість світла [см/с]
k_B   = 1.380649e-16     # стала Больцмана [ерг/К]
m_p   = 1.6726219e-24    # маса протона [г]
M_sun = 1.989e33         # маса Сонця [г]
mu     = 1.0             # середня молекулярна маса

# === ВХІДНІ ПАРАМЕТРИ ===
M_BH   = 21.2             # маса чорної діри [M☉]
T      = 1e8             # температура [К]
Const  = 0.43            # безрозмірна константа для рівняння
r_disk = 100.0          # позиція диска [в одиницях r_g]
r_dyn  = r_disk          # радіус для динамічного часу [в одиницях r_g]

# === 1. ГРАВІТАЦІЙНИЙ РАДІУС ===
M_cgs = M_BH * M_sun
r_g   = 2 * G * M_cgs / c**2     # у сантиметрах
print(f"Гравітаційний радіус: r_g = {r_g:.2e} см")

# === 2. МЕЖІ ДИСКА (розв’язання квадратного рівняння) ===
S = r_disk / Const
P = r_disk**2 / (2 * Const)
D = S**2 - 4 * P

if D < 0:
    raise ValueError("Немає дійсних коренів рівняння для заданих параметрів.")

sqrt_D = np.sqrt(D)
r1 = 0.5 * (S - sqrt_D)
r2 = 0.5 * (S + sqrt_D)

print(f"Межі диска (в r_g): r1 = {r1:.3f}, r2 = {r2:.3f}")

# === 3. ДИНАМІЧНИЙ ЧАС ===
R_cm   = r_dyn * r_g
t_dyn  = 2 * np.pi * np.sqrt(R_cm**3 / (G * M_cgs))     # [с]
v_s    = np.sqrt(k_B * T / (mu * m_p))                  # звукова швидкість [см/с]
t_unit = r_g / v_s                                      # одиниця часу [с]
t_tilde_dyn = t_dyn / t_unit                            # безрозмірний час

print(f"Звукова швидкість: v_s = {v_s:.2e} см/с")
print(f"Фізичний динамічний час: t_dyn = {t_dyn:.2e} с")
print(f"Одиниця часу Athena++: t_unit = {t_unit:.2e} с")
print(f"Безрозмірний динамічний час: t̃_dyn = {t_tilde_dyn:.2e}")
```

```python
import numpy as np

G = 6.67430e-8         # гравітаційна стала, см^3 г^-1 с^-2
k_B = 1.380649e-16     # стала Больцмана, ерг/К
m_p = 1.6726219e-24    # маса протона, г

r = 100
M = 21.2 * 1.989e33           # маса чорної діри, г (масса Сонця)
l0 = 6.26e6              # характерна довжина в моделі, см
T = 1e5                # температура, К
mu = 1               # середня молекулярна маса

t_dyn = 2 * np.pi * np.sqrt((l0*r)**3 / (G * M))

v_s = np.sqrt(k_B * T / (mu * m_p))  # звукова швидкість
t_unit = l0 / v_s

t_tilde_dyn = t_dyn / t_unit

print(f"Звукова швидкість: v_s = {v_s:.2e} см/с")
print(f"Фізичний динамічний час: t_dyn = {t_dyn:.2e} с")
print(f"Одиниця часу Athena++: t_unit = {t_unit:.2e} с")
print(f"Безрозмірний динамічний час: t̃_dyn = {t_tilde_dyn:.2e}")
```

```python
# Гравітаційний радіус в залежності від маси у масах Сонця

mass = 5e6

# Константи
G = 6.67430e-8       # гравітаційна стала [см^3 г^-1 с^-2]
c = 2.99792458e10    # швидкість світла [см/с]
M_sun = 1.989e33     # маса Сонця [г]

def rg_in_cm(mass_in_solar_masses):

    M = mass_in_solar_masses * M_sun
    rg = 2 * G * M / c**2
    return rg

rg_cm = rg_in_cm(mass)
print(f"Гравітаційний радіус для {mass} M☉ = {rg_cm:.2e} см")
```
