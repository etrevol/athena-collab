```python
# Еволюція акреційного диска
# Multiple frames

import numpy as np
import matplotlib.pyplot as plt
import os
from pypalettes import load_cmap

# --- Параметри ---
data_dir = "/home/etrevol/work/acc_diskk/data"
output_dir = "/home/etrevol/results/acc_disk/test/evol"
file_pattern = "acc_disk.block0.out1.{:05d}.tab"
file_prefix = "test"
frames = [0, 1, 10, 100, 1000]
skip = 2

# --- Кольори ---
cmap = load_cmap("Abbott")
colors = [cmap(i) for i in np.linspace(0, 1, len(frames))]

# --- Побудова одного графіка з кількома лініями ---
def plot_quantity(index, ylabel, legend_prefix, filename, title):
    plt.figure(figsize=(7, 3.5))
    for i, frame in enumerate(frames):
        path = os.path.join(data_dir, file_pattern.format(frame))
        data = np.loadtxt(path, skiprows=skip)
        r = data[:, 1]
        y = data[:, index]
        plt.plot(r, y, color=colors[i], label=f"{legend_prefix}{frame}")
    plt.xlabel(r"$r/r_g$")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.legend(title="ОП")
    plt.grid()
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, f"multipleframes_{file_prefix}_{filename}.png"), dpi=200)
    plt.close()

plot_quantity(2, r"$\rho$", "", "density", r"Еволюція густини $ρ$ (ОП $-$ орбітальний період)")
plot_quantity(3, "P", "", "pressure", r"Еволюція тиску $P$ (ОП $-$ орбітальний період)")
plot_quantity(4, r"$v_r / v_s$", "", "vr", r"Еволюція радіальної швидкості $v_r$ (ОП $-$ орбітальний період)")
plot_quantity(5, r"$v_\phi / v_s$", "", "vphi", r"Еволюція азимутальної швидкості $v_\phi$ (ОП $-$ орбітальний період)")

print("Графіки збережено у:", output_dir)
```

```python
# Еволюція акреційного диска
# Subplots

import numpy as np
import matplotlib.pyplot as plt
import os

# --- Параметри ---
data_dir = "/home/etrevol/work/acc_diskk/data"
output_dir = "/home/etrevol/results/acc_disk/test/evol"
file_pattern = "acc_disk.block0.out1.{:05d}.tab"
file_prefix = "test"
frames = [0, 10, 50, 100, 500, 1000]
skip = 2

# --- Функція побудови підграфіків з окремими межами y ---
def plot_subplots(index, ylabel, filename, title, ymin=None, ymax=None, style='k-'):
    # Якщо межі не задані — обчислити автоматично
    if ymin is None or ymax is None:
        y_all = []
        for frame in frames:
            path = os.path.join(data_dir, file_pattern.format(frame))
            data = np.loadtxt(path, skiprows=skip)
            y_all.append(data[:, index])
        y_all = np.concatenate(y_all)
        ymin = np.min(y_all) if ymin is None else ymin
        ymax = np.max(y_all) if ymax is None else ymax

    # Побудова сітки
    n = len(frames)
    ncols = 2
    nrows = (n + ncols - 1) // ncols
    fig, axes = plt.subplots(nrows, ncols, figsize=(12, 8))
    axes = axes.flatten()

    for i, frame in enumerate(frames):
        path = os.path.join(data_dir, file_pattern.format(frame))
        data = np.loadtxt(path, skiprows=skip)
        r = data[:, 1]
        y = data[:, index]
        axes[i].plot(r, y, style)
        axes[i].set_title(f"{frame} ОП")
        axes[i].set_xlabel(r"$r/r_g$")
        axes[i].set_ylabel(ylabel)
        axes[i].set_ylim(ymin, ymax)
        axes[i].grid()

    for j in range(i + 1, len(axes)):
        axes[j].axis('off')

    fig.suptitle(title)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, f"subplots_{file_prefix}_{filename}.png"), dpi=200)
    plt.close()

# --- Побудова графіків з класичними стилями ---
plot_subplots(2, r"$\rho$", "density", "Еволюція густини $ρ$ (ОП $-$ орбітальний період)", ymin=0, ymax=1.1, style='r-')
plot_subplots(3, "P", "pressure", "Еволюція тиску $P$ (ОП $-$ орбітальний період)", ymin=0, ymax=1600, style='g-')
plot_subplots(4, r"$v_r / v_s$", "vr", "Еволюція радіальної швидкості $v_r$ (ОП $-$ орбітальний період)", style='b-')
plot_subplots(5, r"$v_\phi / v_s$", "vphi", r"Еволюція азимутальної швидкості $v_\phi$ (ОП $-$ орбітальний період)", style='m-')


print("Графіки збережено у:", output_dir)
```
